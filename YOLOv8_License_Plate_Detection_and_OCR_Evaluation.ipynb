```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
from google.colab.patches import cv2_imshow

def preprocess_image(image_path):
    # Read the image
    image = cv2.imread(image_path)
    if image is None:
        raise FileNotFoundError(f"Image at path {image_path} not found.")

    # Convert to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Noise reduction with Gaussian blur
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)

    # Apply binary thresholding
    _, binary = cv2.threshold(blurred, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

    # Apply morphological operations
    kernel = np.ones((3, 3), np.uint8)
    morph = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)

    # Optionally, use edge detection (Canny)
    edges = cv2.Canny(morph, 100, 200)

    # Display the results
    titles = ['Original Image', 'Gray Image', 'Blurred Image', 'Binary Image', 'Morphological Operations', 'Edges']
    images = [image, gray, blurred, binary, morph, edges]

    for i in range(6):
        plt.subplot(2, 3, i+1)
        plt.imshow(images[i], 'gray')
        plt.title(titles[i])
        plt.xticks([]), plt.yticks([])

    plt.show()

    # Save the final preprocessed image
    preprocessed_image_path = 'preprocessed_license_plate.png'
    cv2.imwrite(preprocessed_image_path, morph)

    return preprocessed_image_path

# Input box for the user to specify the image path
image_path = input("Please enter the path to the license plate image: ")

# Preprocess the image and display/save the result
try:
    preprocessed_image_path = preprocess_image(image_path)
    print(f"Preprocessed image saved at {preprocessed_image_path}")
except FileNotFoundError as e:
    print(e)
```

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

def preprocess_image(image_path):
    # Read the image
    image = cv2.imread(image_path)
    if image is None:
        raise FileNotFoundError(f"Image at path {image_path} not found.")

    # Convert to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Noise reduction with Gaussian blur
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)

    # Apply binary thresholding
    _, binary = cv2.threshold(blurred, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

    # Apply morphological operations
    kernel = np.ones((3, 3), np.uint8)
    morph = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)

    # Display the selected results in one image
    titles = ['Gray Image', 'Blurred Image', 'Binary Image', 'Morphological Operations']
    images = [gray, blurred, binary, morph]

    plt.figure(figsize=(12, 8))
    for i in range(4):
        plt.subplot(2, 2, i+1)
        plt.imshow(images[i], cmap='gray')
        plt.title(titles[i])
        plt.axis('off')

    plt.tight_layout()
    plt.show()

    # Save the final preprocessed image
    preprocessed_image_path = 'preprocessed_license_plate.png'
    cv2.imwrite(preprocessed_image_path, morph)

    return preprocessed_image_path

# Input box for the user to specify the image path
image_path = input("Please enter the path to the license plate image: ")

# Preprocess the image and display/save the result
try:
    preprocessed_image_path = preprocess_image(image_path)
    print(f"Preprocessed image saved at {preprocessed_image_path}")
except FileNotFoundError as e:
    print(e)
```

```python
# Import necessary libraries
import easyocr
from paddleocr import PaddleOCR
import pytesseract
from PIL import Image
import numpy as np
import pandas as pd
import os

# Initialize OCR tools
paddle_ocr = PaddleOCR(use_angle_cls=True, lang='en')  # PaddleOCR for English
easy_ocr = easyocr.Reader(['en'])  # EasyOCR for English

# Define the function to get OCR results with PaddleOCR
def get_paddle_text_and_confidence(image_path):
    paddle_result = paddle_ocr.ocr(image_path, cls=True)
    paddle_text = " ".join([word[1][0] for line in paddle_result for word in line])
    paddle_confidences = [word[1][1] for line in paddle_result for word in line]
    return paddle_text, paddle_confidences

# Define the function to get OCR results with EasyOCR
def get_easyocr_text_and_confidence(image_path):
    easyocr_result = easy_ocr.readtext(image_path)
    easyocr_text = " ".join([res[1] for res in easyocr_result])
    easyocr_confidences = [res[2] for res in easyocr_result]
    return easyocr_text, easyocr_confidences

# Define the function to get OCR results with Tesseract
def get_tesseract_text_and_confidence(image_path):
    tesseract_data = pytesseract.image_to_data(Image.open(image_path), output_type=pytesseract.Output.DATAFRAME)
    tesseract_words = tesseract_data[tesseract_data['text'].notna()]
    if not tesseract_words.empty:
        tesseract_text = " ".join(tesseract_words['text'])
        tesseract_confidences = [conf / 100 for conf in tesseract_words['conf']]
    else:
        tesseract_text = "No text detected"
        tesseract_confidences = []
    return tesseract_text, tesseract_confidences

# Function to calculate mean and median confidence
def calculate_confidence(confidence_list):
    if not confidence_list:
        mean_confidence = median_confidence = 0  # Set to 0 when list is empty
    else:
        mean_confidence = np.mean(confidence_list)
        median_confidence = np.median(confidence_list)
    return mean_confidence, median_confidence

# Get the path to the folder containing the images
folder_path = input("Enter the folder path containing the images: ")

# Check if the folder exists
if not os.path.exists(folder_path):
    raise ValueError("The specified folder does not exist.")

# Get a list of all image files in the folder
image_files = [f for f in os.listdir(folder_path) if f.lower().endswith((".jpg", ".jpeg", ".png"))]

# Loop over each image file and perform OCR
ocr_results = {}
for image_file in image_files:
    image_path = os.path.join(folder_path, image_file)
    # Get OCR results and confidence levels for each tool
    paddle_text, paddle_confidences = get_paddle_text_and_confidence(image_path)
    easyocr_text, easyocr_confidences = get_easyocr_text_and_confidence(image_path)
    tesseract_text, tesseract_confidences = get_tesseract_text_and_confidence(image_path)

    # Calculate mean and median confidence for each OCR tool
    paddle_mean, paddle_median = calculate_confidence(paddle_confidences)
    easyocr_mean, easyocr_median = calculate_confidence(easyocr_confidences)
    tesseract_mean, tesseract_median = calculate_confidence(tesseract_confidences)

    # Store OCR results for this image
    ocr_results[image_file] = {
        "PaddleOCR": (paddle_text, paddle_mean, paddle_median),
        "EasyOCR": (easyocr_text, easyocr_mean, easyocr_median),
        "Tesseract": (tesseract_text, tesseract_mean, tesseract_median),
    }

# Display OCR results for all images
for image_file, results in ocr_results.items():
    print(f"\nOCR Results for {image_file}:")
    for key, value in results.items():
        text, mean_confidence, median_confidence = value
        print(f"{key}:")
        print("  Text:", text)
        print("  Confidence:", mean_confidence)
```

```python
nimport matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Function to calculate confidence with scaling
def scale_confidence(confidence_list, scale=1):
    if not confidence_list:
        return None
    mean_confidence = np.mean(confidence_list) / scale
    return mean_confidence

# Create a DataFrame for easy plotting
image_files = list(ocr_results.keys())
df = pd.DataFrame({
    "Image": image_files,
    "PaddleOCR Confidence": [
        scale_confidence(ocr_results[img]["PaddleOCR"][1], 1) for img in image_files
    ],
    "EasyOCR Confidence": [
        scale_confidence(ocr_results[img]["EasyOCR"][1], 1) for img in image_files
    ],
    "Tesseract Confidence": [
        scale_confidence(ocr_results[img]["Tesseract"][1], 1) for img in image_files
    ],
})

# Plot the data
plt.figure(figsize=(12, 6))
sns.barplot(data=df.melt(id_vars=["Image"], var_name="OCR Tool", value_name="Mean Confidence"),
            x="Image", y="Mean Confidence", hue="OCR Tool")
plt.xlabel("Image")
plt.ylabel("Confidence (Scaled to 0-1)")
plt.title("Confidence of OCR Across Images")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
